```python
import math

# Entrada de datos
r = float(input("Ingrese el radio (m): "))
h = float(input("Ingrese la altura (m): "))
densidad = 0.75

# Proceso
volumen = math.pi * (r**2) * h
toneladas = volumen * densidad

# Salida
print(f"Resultado: {toneladas:.2f} toneladas")
```

```python
# --- PROGRAMA DE DOSIFICACIÓN DE FERTILIZANTE ---
# Estudiante: DANNA PAOLA

import math

def calcular_dosificacion():
    print("--- SISTEMA DE RIEGO POR GOTEO: CÁLCULO DE MEZCLA ---")

    try:
        #  ENTRADA (E)
        hectareas = float(input("Ingrese las hectáreas totales (ha): "))
        dosis_por_ha = float(input("Ingrese la dosis requerida (L/ha): "))
        relacion_agua = 500  # 1 litro de producto por 500 de agua

        #  PROCESO (P)
        # 1. Calculamos el total de producto puro
        total_producto = hectareas * dosis_por_ha

        # 2. Calculamos el agua necesaria basada en la relación 1:500
        total_agua = total_producto * relacion_agua

        # --- SALIDA (S) ---
        print("\n" + "="*30)
        print(f"RESUMEN PARA {hectareas} HECTÁREAS:")
        print(f"-> Fertilizante puro necesario: {total_producto:.2f} Litros")
        print(f"-> Agua para dilución: {total_agua:.2f} Litros")
        print(f"-> Volumen total de la mezcla: {total_producto + total_agua:.2f} Litros")
        print("="*30)

    except ValueError:
        print("Error: Por favor, introduce valores numéricos (usa punto para decimales).")

# Ejecutar la función
calcular_dosificacion()
```

```python
# --- CÁLCULO DE RENDIMIENTO DE PULPA EN FRUTAS ---
# Estudiante: DANNA PAOLA
# Ingeniería Agroindustrial

def calcular_rendimiento():
    print("="*40)
    print("SISTEMA DE CONTROL DE RENDIMIENTO - PLANTA")
    print("="*40)

    try:
        # --- ENTRADA (E) ---
        peso_total = float(input("Ingrese el peso total de la fruta comprada (kg): "))
        peso_residuos = float(input("Ingrese el peso de los residuos (kg): "))

        # --- PROCESO (P) ---
        if peso_residuos > peso_total:
            print("\n[ERROR]: Los residuos no pueden ser mayores al peso total.")
        else:
            # Cálculo de pulpa aprovechable
            pulpa = peso_total - peso_residuos

            # Cálculo del porcentaje de rendimiento
            rendimiento = (pulpa / peso_total) * 100

            # --- SALIDA (S) ---
            print("\n" + "-"*30)
            print("REPORTE DE APROVECHAMIENTO")
            print("-"*30)
            print(f"Pulpa obtenida:      {pulpa:.2f} kg")
            print(f"Residuos (merma):    {peso_residuos:.2f} kg")
            print(f"Rendimiento final:   {rendimiento:.2f}%")
            print("-"*30)

            # Nota técnica según el resultado
            if rendimiento < 50:
                print("Nota: El rendimiento es bajo. Revisar calidad del lote.")
            else:
                print("Nota: Rendimiento dentro de los parámetros normales.")

    except ValueError:
        print("\n[ERROR]: Ingrese solo números. Use punto (.) para decimales.")
    except ZeroDivisionError:
        print("\n[ERROR]: El peso total no puede ser cero.")

# Ejecutar el programa
calcular_rendimiento()
```

```python
# --- CÁLCULO DE DOSIFICACIÓN DE CEMENTO ---
# Estudiante: DANNA PAOLA

# 1. ENTRADA (E)
area_muro = float(input("Ingrese el área del muro (m2): "))
factor_cemento = 0.25  # Bultos por cada m2

# 2. PROCESO (P)
bultos_necesarios = area_muro * factor_cemento

# 3. SALIDA (S)


print(f"\n--- REPORTE DE MATERIALES ---")
print(f"Para un muro de {area_muro} m2 se requieren:")
print(f"-> {bultos_necesarios:.2f} bultos de cemento.")
```

```python
# --- CÁLCULO DE PESO DE VIGA DE ACERO ---
# Estudiante: DANNA PAOLA

# 1. ENTRADA (E)
base = float(input("Ingrese la base de la viga (m): "))
altura = float(input("Ingrese la altura de la viga (m): "))
longitud = float(input("Ingrese la longitud de la viga (m): "))
densidad_acero = 7850  # kg/m^3

# 2. PROCESO (P)
peso_viga = base * altura * longitud * densidad_acero

# 3. SALIDA (S)
print(f"\n--- REPORTE ESTRUCTURAL ---")
print(f"El peso propio de la viga es: {peso_viga:.2f} kg")
```

```python
# --- CÁLCULO DE PENDIENTE DE RAMPA ---
# Estudiante: DANNA PAOLA

# 1. ENTRADA (E)
altura = float(input("Ingrese la altura a vencer (m): "))
longitud = float(input("Ingrese la longitud horizontal (m): "))

# 2. PROCESO (P)
porcentaje_pendiente = (altura / longitud) * 100

# 3. SALIDA (S)
print(f"\n--- ANÁLISIS DE ACCESIBILIDAD ---")
print(f"La pendiente de la rampa es: {porcentaje_pendiente:.2f}%")

# Condicional extra para normativa (Opcional)
if porcentaje_pendiente <= 8:
    print("Estado: CUMPLE la normativa estándar.")
else:
    print("Estado: NO CUMPLE (Muy inclinada para accesibilidad).")
```
