# **Control de Rendimiento en Extracción de Aceite**
Este ejercicio es ideal para introducir float, int y la concatenación de strings.

**Planteamiento:**

Una planta procesadora de palma de aceite necesita calcular **el Índice de Rendimiento de Extracción (IRE).** El operario debe ingresar **el peso total de los racimos de fruta fresca (RFF)** procesados en kilogramos y la cantidad de aceite crudo obtenido en litros.

**Requerimientos:**

1.Solicitar el nombre del lote de producción (Tipo string).

2.Solicitar el peso de los racimos en kg (Tipo float).(pr)

3.Solicitar los litros de aceite obtenidos (Tipo float).(lt)

4.Calcular el rendimiento sabiendo que la densidad aproximada del aceite es de 0.92 kg/L.(r)

$$pa(pa)=L*0.92$$
$$r=(pa/pr)*100$$

**Prueba del programa:**

Si se ingresan 1000 kg de racimos y se obtienen 200 litros de aceite, el programa debe informar que el rendimiento del lote es del 18.4%.







```python
#ENTRADA:Solicitar el nombre del lote de producción (Tipo string).
#Solicitar el peso de los racimos en kg (Tipo float).(pr)
#Solicitar los litros de aceite obtenidos (Tipo float).(lt)
nombre=input("Ingrese el nombre del lote: ")
pr=float(input("Ingrese el peso de los racimos en kg: "))
lt=float(input("Ingrese los litros de aceite obtenidos: "))

#PROCESO CALCULAR PESO DE ACEITE(pa) Y EL REDIMIENTO (r)
pa=lt*0.92
r=(pa/pr)*100

#SALIDA: Informar el rendimiento del lote.
print("El rendimiento del lote",nombre,"es del",r,"%")






```

Aquí está la explicación línea por línea del código proporcionado:

*   `#ENTRADA:Solicitar el nombre del lote de producción (Tipo string).`
    *   Este es un comentario que indica la sección de entrada del programa y lo que se solicitará (el nombre del lote como un string).
*   `#Solicitar el peso de los racimos en kg (Tipo float).(pr)`
    *   Otro comentario que especifica la solicitud del peso de los racimos en kilogramos (tipo flotante) que se almacenará en la variable `pr`.
*   `#Solicitar los litros de aceite obtenidos (Tipo float).(lt)`
    *   Comentario que describe la solicitud de los litros de aceite obtenidos (tipo flotante) para la variable `lt`.
*   `nombre=input("Ingrese el nombre del lote: ")`
    *   Esta línea solicita al usuario que ingrese el nombre del lote de producción. El texto "Ingrese el nombre del lote: " se muestra en la consola. La entrada del usuario se guarda como una cadena de texto en la variable `nombre`.
*   `pr=float(input("Ingrese el peso de los racimos en kg: "))`
    *   Esta línea solicita al usuario el peso de los racimos en kilogramos. El texto "Ingrese el peso de los racimos en kg: " se muestra en la consola. La entrada del usuario, que es una cadena de texto, se convierte a un número de punto flotante (`float`) y se almacena en la variable `pr`.
*   `lt=float(input("Ingrese los litros de aceite obtenidos: "))`
    *   Similar a la anterior, esta línea solicita los litros de aceite obtenidos. La entrada del usuario se convierte a un número de punto flotante (`float`) y se guarda en la variable `lt`.
*   `#PROCESO CALCULAR PESO DE ACEITE(pa) Y EL REDIMIENTO (r)`
    *   Comentario que marca el inicio de la sección de procesamiento, donde se calculará el peso del aceite (`pa`) y el rendimiento (`r`).
*   `pa=lt*0.92`
    *   Esta línea calcula el peso del aceite (`pa`). Multiplica los litros de aceite obtenidos (`lt`) por la densidad del aceite (0.92 kg/L). El resultado se almacena en la variable `pa`.
*   `r=(pa/pr)*100`
    *   Esta línea calcula el rendimiento de extracción (`r`). Divide el peso del aceite (`pa`) por el peso de los racimos (`pr`) y luego multiplica el resultado por 100 para obtener un porcentaje. El valor se guarda en la variable `r`.
*   `#SALIDA: Informar el rendimiento del lote.`
    *   Comentario que indica la sección de salida del programa.
*   `print("El rendimiento del lote",nombre,"es del",r,"%")`
    *   Esta línea imprime el resultado final al usuario. Construye una cadena de texto concatenando varias partes: "El rendimiento del lote", el nombre del lote (`nombre`), "es del", el valor del rendimiento (`r`), y finalmente "%". Esto muestra el rendimiento calculado de una manera legible.
